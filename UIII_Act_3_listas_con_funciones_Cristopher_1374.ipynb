```python
# Sistema de gestión de Ilustradores utilizando funciones y while

ilustradores = []

# Función para agregar ilustrador
def agregarIlustrador():
    nombre = input("Nombre del ilustrador: ")
    estilo = input("Estilo de dibujo: ")

    ilustrador = {
        "nombre": nombre,
        "estilo": estilo
    }

    ilustradores.append(ilustrador)
    print("Ilustrador agregado correctamente.\n")


# Función para listar ilustradores
def listarIlustradores():
    if len(ilustradores) == 0:
        print("No hay ilustradores registrados.\n")
    else:
        print("\nLista de ilustradores:")
        contador = 1

        while contador <= len(ilustradores):
            ilu = ilustradores[contador - 1]
            print(contador, "-", ilu["nombre"], "-", ilu["estilo"])
            contador += 1

        print()


# Función para buscar ilustrador
def buscarIlustrador():
    nombre = input("Ingrese el nombre a buscar: ")
    encontrado = False
    contador = 0

    while contador < len(ilustradores):
        if ilustradores[contador]["nombre"] == nombre:
            print("Ilustrador encontrado:")
            print("Nombre:", ilustradores[contador]["nombre"])
            print("Estilo:", ilustradores[contador]["estilo"])
            encontrado = True

        contador += 1

    if encontrado == False:
        print("Ilustrador no encontrado.\n")


# Función para modificar ilustrador
def modificarIlustrador():
    nombre = input("Nombre del ilustrador a modificar: ")
    contador = 0
    encontrado = False

    while contador < len(ilustradores):
        if ilustradores[contador]["nombre"] == nombre:

            nuevoNombre = input("Nuevo nombre: ")
            nuevoEstilo = input("Nuevo estilo: ")

            ilustradores[contador]["nombre"] = nuevoNombre
            ilustradores[contador]["estilo"] = nuevoEstilo

            print("Ilustrador modificado correctamente.\n")
            encontrado = True

        contador += 1

    if encontrado == False:
        print("Ilustrador no encontrado.\n")


# Función para borrar ilustrador
def borrarIlustrador():
    nombre = input("Nombre del ilustrador a borrar: ")
    contador = 0
    encontrado = False

    while contador < len(ilustradores):

        if ilustradores[contador]["nombre"] == nombre:
            ilustradores.pop(contador)
            print("Ilustrador eliminado correctamente.\n")
            encontrado = True
            break

        contador += 1

    if encontrado == False:
        print("Ilustrador no encontrado.\n")


# Menú principal
opcion = 0

while opcion != 6:

    print("===== SISTEMA DE ILUSTRADORES =====")
    print("1. Agregar ilustrador")
    print("2. Listar ilustradores")
    print("3. Buscar ilustrador")
    print("4. Modificar ilustrador")
    print("5. Borrar ilustrador")
    print("6. Salir")

    opcion = int(input("Seleccione una opción: "))

    if opcion == 1:
        agregarIlustrador()

    elif opcion == 2:
        listarIlustradores()

    elif opcion == 3:
        buscarIlustrador()

    elif opcion == 4:
        modificarIlustrador()

    elif opcion == 5:
        borrarIlustrador()

    elif opcion == 6:
        print("Saliendo del sistema...")

    else:
        print("Opción inválida.\n")
```
